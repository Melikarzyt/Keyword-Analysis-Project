Step 1:Import libraries

```python
import pandas as pd
import numpy as np
```

Step 2:Load the dataset

```python
df = pd.read_csv('yektanet.csv', encoding='utf8')
df.tail()
```

Step 3: We group the data by the keywords and count how many times each keyword appears in the dataset.
This helps us undrestand which keywords are more frequent.

```python
grouped = df.groupby('keywords')
keyword_counts = grouped['keywords'].apply(len).reset_index(name='count')
keyword_counts.head()
```

Step 4: Filter frequent keywords

```python
filtered_df = keyword_counts[keyword_counts['count'] > 1400]
filtered_df
```

Step 5: Calculated the popularity column

We create a new column called 'popularity' by dividing each rows 'visit count' by the total 'visit count' of its keyword group.

```python
df['popularity'] = df['visit count'] / df.groupby('keywords')['visit count'].transform('sum') * 100
df['popularity'] = df['popularity'].round(2)
```
